## Task 3 — Define Your Synthetic Well-Test Experiment

```python
pi = 4000 # psi
q = 500 # STB/s
k = 50 # md
h = 50 # ft
ϕ = 0.2
μ = 1.0 # cP
B = 1.2 # RB/STB
ct = 1.5 * 10**(-5) # psi-1
rw = 0.25 # ft
```

## Task 4 Line Source Solution

### Task 4A — Understand Dimensionless Time

```python
tD = (0.0002637 * k * time_hr) / (ϕ * μ * ct * rw**2)
tD
```

```python
print(len(tD))
print(np.min(tD))
print(np.max(tD))
print(tD[:5])
```

### Task 4B — Use the Line Source Solution

```python
pD = -0.5 * expi(-1 / (4 * tD))
pD
```

```python
pD = -0.5 * expi(-1 / (4 * tD))

print(len(pD))
print(np.min(pD))
print(np.max(pD))
print(pD[:5])
```

### Task 4C — Convert \(p_D\) into Actual Pressure Drop

```python
delta_p = (141.2 * q * μ * B * pD) / (k * h)

pwf = pi - delta_p
```

```python
print(len(delta_p))
print(delta_p[:5])

print(len(pwf))
print(pwf[:5])

print("Minimum flowing pressure:", np.min(pwf))
print("Maximum flowing pressure:", np.max(pwf))
```

# Week 1B : Using welltestpy

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.special import expi
import welltestpy as wp
```

## 1. Petroleum reservoir properties

```python
k_md = 50.0          # permeability, mD
h_ft = 50.0          # reservoir thickness, ft
phi = 0.20           # porosity, fraction
mu_cp = 1.0          # viscosity, cP
ct_psi = 1.5e-5      # total compressibility, psi^-1
rho = 800.0          # oil density, kg/m^3
B = 1.2              # formation volume factor, RB/STB
q_stb_day = 500.0    # production rate, STB/day
rw = 0            # production well radius, ft
r_m  = 100           # observation well distance, m
g = 9.80665          # m/s^2
```

## 2. Convert petroleum properties to SI units

```python
k = k_md * 9.869233e-16       # mD -> m^2

h = h_ft * 0.3048             # ft -> m

mu = mu_cp * 1e-3             # cP -> Pa.s

ct = ct_psi / 6894.757        # psi^-1 -> Pa^-1
```

## 3. Calculate equivalent Theis parameters

```python
# Transmissivity, m^2/s
transmissivity = (k * rho * g * h) / mu

# Storage coefficient, dimensionless
storage = rho * g * phi * ct * h


print("Transmissivity =", transmissivity, "m^2/s")
print("Storage coefficient =", storage)
```

## 4. Convert petroleum production rate to m^3/s

```python
# Surface rate -> reservoir rate
q_res_bbl_day = q_stb_day * B

# Reservoir bbl/day -> m^3/s
Q = q_res_bbl_day * 0.1589873 / 86400

# Negative because production/extraction
rate = -Q

print("Production rate =", rate, "m^3/s")
```

## 5. Create WellTestPy campaign

```python
campaign = wp.data.Campaign(
    name="Petroleum_Analogue_Campaign",
    description="Petroleum radial-flow analogue using Theis solution"
)
```

## 6. Add production and observation wells

```python
if rw > 0:
    r_m = 0.25 * 0.3048
    
campaign.add_well(
    name="PW1",
    radius=0.0762,
    coordinates=np.array([0.0, 0.0])
)


campaign.add_well(
    name="OW1",
    radius=0.0762,
    coordinates=np.array([r_m, 0.0])
)
```

```python
campaign.wells
```

```python
campaign.plot_wells()
plt.savefig(f"Well Top Field View @ {r_m}m.jpeg", dpi=300, bbox_inches="tight")
```

## 7. Create pumping test

```python
test = wp.data.PumpingTest(
    name="Test1",
    pumpingwell="PW1",
    pumpingrate=rate
)

campaign.addtests({
    "Test1": test
})
```

## 8. Create time array

```python
# 0.001 hr to 100 hr
time_hr = np.logspace(-3, 2, 100)

# WellTestPy requires seconds
time_sec = time_hr * 3600

rad_array = np.array([r_m])

# Distance from PW1 to OW1
# rad_array = np.array([100.0])
```

## 9. Run Theis analytical solution

```python
head = wp.estimate.transient_lib.ana.theis(
    time=time_sec,
    rad=rad_array,
    storage=storage,
    transmissivity=transmissivity,
    rate=rate
)

print("Head array shape:", head.shape)
```

## 10. Convert head response to positive drawdown

```python
head_array = head.flatten()

drawdown_m = -head_array
```

## 11. Convert drawdown from meters of head to psi

```python
delta_p_pa = rho * g * drawdown_m

delta_p_psi = delta_p_pa / 6894.757
```

## 12. Inspect results

```python
print("\nFirst 5 times (hr):")
print(time_hr[:5])

print("\nFirst 5 pressure drops (psi):")
print(delta_p_psi[:5])

print("\nLast 5 pressure drops (psi):")
print(delta_p_psi[-5:])
```

## 13. Convert drawdown from meters of head to psi

```python
pi = 4000.0  # psi

pwf_psi = pi - delta_p_psi

print("\nFirst 5 pressure drops (psi):")
print(pwf_psi[:5])

print("\nLast 5 pressure drops (psi):")
print(pwf_psi[-5:])
```

## 14. Build the Pandas DataFrame

```python
df_wpy = pd.DataFrame({
    "Time_hr":time_hr,
    "DeltaP_psi": delta_p_psi,
    "Pwf_psi": pwf_psi
})
```
